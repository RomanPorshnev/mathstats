```python
import numpy as np
from scipy.optimize import minimize


def weibull_pdf(x, params):
    k, lam = params
    return (k / lam) * (x / lam)**(k - 1) * np.exp(-(x / lam)**k)
def negative_log_likelihood(params, data, pdf):
    return -np.sum([np.log(pdf(x, params)) for x in data])
def MLE(data, start_params, pdf, bounds):
    result = minimize(fun=negative_log_likelihood, x0=start_params, args=(data, pdf), method='L-BFGS-B', bounds=[(None, None), (1e-06, None)])
    return result.x
```

```python
lam, k = 5, 2
data = lam * np.random.weibull(a=k, size=10000)
params = MLE(data=data, start_params=(0.1, 0.1), pdf=weibull_pdf, bounds=[(1e-06, None), (1e-06, None)])
print(f"k = {params[0]:.3f}, lamda = {params[1]:.3f}")
```
