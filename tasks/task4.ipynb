```python
import numpy as np
from scipy.stats import skew, kurtosis

x = np.random.normal(loc=2, size=10000)

print(f'E[X] = {np.mean(x)}')
print(f'median[X] = {np.median(x)}')

print(f'Var[X] (biased) = {np.var(x, ddof=0)}')
print(f'Var[X] (unbiased) = {np.var(x, ddof=1)}')

print(f'Std[X] (biased) = {np.std(x, ddof=0)}')
print(f'Std[X] (unbiased) = {np.std(x, ddof=1)}')

print(f'Sk[X] = {skew(x, bias=False)}')
print(f'Ku[X] = {kurtosis(x, fisher=True, bias=False)}')
```
