```python
import numpy as np
from scipy.stats import skew, kurtosis

data = np.random.normal(loc=2, size=10000)

print(f'E[X] = {np.mean(data)}')
print(f'median[X] = {np.median(data)}')

print(f'Var[X] (biased) = {np.var(data, ddof=0)}')
print(f'Var[X] (unbiased) = {np.var(data, ddof=1)}')

print(f'Std[X] (biased) = {np.std(data, ddof=0)}')
print(f'Std[X] (unbiased) = {np.std(data, ddof=1)}')

print(f'Sk[X] = {skew(data, bias=False)}')
print(f'Ku[X] = {kurtosis(data, fisher=True, bias=False)}')
```
