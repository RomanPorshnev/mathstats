```python
import numpy as np
import seaborn as sns

sns.set_style("whitegrid")
data = np.random.normal(size=1000)
ax = sns.histplot(data, stat='density', kde=True)
```

## Замечание

Ширина ядра в классе `scipy.stats.gaussian_kde(...)`, который вызывается в `seaborn.histplot(kde=True, ...)`определяется следующим образом:

$$\sigma_{ker} = \sigma_{X} \cdot n^{-1/5}$$

где:
- $\sigma_{X}$ — стандартное отклонение выборки;
- $n$ — объём выборки.
